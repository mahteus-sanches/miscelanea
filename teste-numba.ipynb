```python
import numba
import numpy as np
import matplotlib
import random
import time
```

```python
random.seed(time.time())
size = 1000
```

```python
def create_random_image():
    vector = np.zeros((size,size,size), dtype=float)
    for i in range(1,size):
        for j in range(1,size):
            for k in range(1,size):
                vector[i,j,k] = random.randrange(256)
    return vector
```

```python
@numba.njit
def create_random_image_numba():
    vector = np.zeros((size,size,size), dtype=float)
    for i in range(1,size):
        for j in range(1,size):
            for k in range(1,size):
                vector[i,j,k] = random.randrange(256)
    return vector
```

```python
start_normal = time.time()
Image1 = create_random_image()
stop_normal = time.time()

start_numba = time.time()
Image2 = create_random_image_numba()
stop_numba = time.time()
```

```python
print('sem numba:', (stop_normal-start_normal))

print(f'com numba:', (stop_numba-start_numba))
```
